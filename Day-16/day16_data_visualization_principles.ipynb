# Day 16 – Data Visualization Principles
## Why Good Charts Create Better Decisions

```
import pandas as pd
import matplotlib.pyplot as plt
```

```
df=pd.read_csv("data/nyc_311_sample.csv")
df.head()
```

```
df["Created Date"]=pd.to_datetime(df["Created Date"])
df["Month"]=df["Created Date"].dt.to_period("M")
```

```
monthly=df.groupby("Month").size()
monthly.plot(figsize=(10,5),marker="o")
plt.title("Monthly Complaints")
plt.show()
```

```
top=df["Complaint Type"].value_counts().head(10)
top.plot.bar(figsize=(10,5))
plt.show()
```

```
borough=df["Borough"].value_counts()
borough.plot.bar(figsize=(8,5))
plt.show()
```

```
df["Resolution Hours"].plot.hist(bins=20)
plt.show()
```

```
df.boxplot(column="Resolution Hours")
plt.show()
```

```
plt.scatter(df["Resolution Hours"],df.index,alpha=0.4)
plt.show()
```
