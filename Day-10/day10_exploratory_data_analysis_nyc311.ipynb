This notebook demonstrates how analysts explore datasets to identify structure, patterns, anomalies, and early analytical signals using the **NYC 311 Service Requests dataset**.

## Step 1 – Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Step 2 – Load Dataset
Replace the file path if your dataset is stored elsewhere.

```python
df = pd.read_csv('nyc_311_sample.csv')
df.head()
```

## Step 3 – Inspect Dataset Size

```python
df.shape
```

## Step 4 – Inspect Dataset Structure

```python
df.info()
```

## Step 5 – View Column Names

```python
df.columns
```

## Step 6 – Explore Complaint Category Distribution

```python
df['Complaint Type'].value_counts().head(10)
```

## Step 7 – Visualize Top Complaint Categories

```python
df['Complaint Type'].value_counts().head(10).plot(kind='bar')
plt.title('Top Complaint Categories')
plt.ylabel('Count')
plt.xlabel('Complaint Type')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Step 8 – Explore Borough Distribution

```python
df['Borough'].value_counts()
```

## Step 9 – Visualize Borough Complaint Counts

```python
df['Borough'].value_counts().plot(kind='bar')
plt.title('Complaints by Borough')
plt.ylabel('Count')
plt.xlabel('Borough')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Step 10 – Convert Date Field

```python
df['Created Date'] = pd.to_datetime(df['Created Date'], errors='coerce')
```

## Step 11 – Create Year Column and Explore Trend

```python
df['Year'] = df['Created Date'].dt.year
df['Year'].value_counts().sort_index()
```

## Step 12 – Visualize Complaint Trends Over Time

```python
df['Year'].value_counts().sort_index().plot()
plt.title('Complaint Reports Over Time')
plt.ylabel('Number of Complaints')
plt.xlabel('Year')
plt.tight_layout()
plt.show()
```

## Step 13 – Cross Analysis
This helps explore how complaint types vary across boroughs.

```python
pd.crosstab(df['Borough'], df['Complaint Type']).head()
```

## Key Takeaways

- EDA helps analysts understand what data contains before formal modeling
- Frequency analysis reveals dominant operational patterns
- Time analysis exposes trend behavior
- Geographic exploration shows concentration patterns
- Simple charts often reveal the most important early insights
