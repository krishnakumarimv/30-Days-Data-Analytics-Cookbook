# Day 7 – Analytics Cookbook
## Common Mistakes Beginners Make in Data Analytics
### Demonstrations Using NYC 311 Service Requests Dataset

This notebook illustrates frequent beginner mistakes in data analytics using the NYC 311 Service Requests dataset. The emphasis is on analytical discipline and reasoning.

## Environment Setup

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Load Dataset

```
df = pd.read_csv("311_Service_Requests.csv", low_memory=False)
df.shape
```

Preview Data

```
df.head()
```

## Mistake 1: Starting Without a Clear Question

```
df["Complaint Type"].value_counts().head()
```

## Mistake 2: Ignoring Data Types

```
df["Created Date"].dtype
```

Correct Date Parsing

```
df["Created Date"] = pd.to_datetime(df["Created Date"], errors="coerce")
```

## Mistake 3: Ignoring Missing Values

```
df.isna().sum().sort_values(ascending=False).head()
```

Cleaning Critical Fields

```
df_clean = df.dropna(subset=["Created Date", "Complaint Type"])
```

## Mistake 4: Duplicate Records

```
df_clean.duplicated().sum()
```

Removing Duplicates

```
df_clean = df_clean.drop_duplicates()
```

## Mistake 5: Incorrect Aggregation

```
monthly_wrong = df_clean.groupby(df_clean["Created Date"].dt.month).size()
monthly_wrong.head()
```

Correct Time Aggregation

```
df_clean["YearMonth"] = df_clean["Created Date"].dt.to_period("M")
monthly_counts = df_clean.groupby("YearMonth").size()
monthly_counts.head()
```

## Mistake 6: Weak Visualization

```
monthly_counts.plot()
plt.show()
```

Improved Visualization

```
monthly_counts.plot(figsize=(12,5))
plt.title("Monthly Complaint Volume")
plt.show()
```

## Mistake 7: Category Dominance vs Trends

```
top_types = df_clean["Complaint Type"].value_counts().head(5)
top_types
```

Descriptive Statistics Before Complexity

```
monthly_counts.describe()
```

## Key Takeaways
Analytics quality depends more on reasoning discipline than tools or algorithms.
