## How Raw Data Becomes Analytical Value

Part of **30 Days of Data Analytics – From Foundations to Decision Intelligence**

This notebook demonstrates how raw operational data can be transformed into more useful analytical features using the **NYC 311 Service Requests dataset**.

## Step 1 – Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Step 2 – Load Dataset

```python
df = pd.read_csv('nyc_311_sample.csv')
df.head()
```

## Step 3 – Convert Date Fields

```python
df['Created Date'] = pd.to_datetime(df['Created Date'], errors='coerce')
df['Closed Date'] = pd.to_datetime(df['Closed Date'], errors='coerce')
```

## Step 4 – Create Time-Based Features

```python
df['Year'] = df['Created Date'].dt.year
df['Month'] = df['Created Date'].dt.month
df['Day'] = df['Created Date'].dt.day
df['Hour'] = df['Created Date'].dt.hour
df['Weekday'] = df['Created Date'].dt.day_name()
```

## Step 5 – Create Resolution Time Feature

```python
df['Resolution Hours'] = (df['Closed Date'] - df['Created Date']).dt.total_seconds() / 3600
```

## Step 6 – Create Open Case Flag

```python
df['Is Open'] = df['Closed Date'].isna().astype(int)
```

## Step 7 – Group Rare Complaint Categories

```python
top_categories = df['Complaint Type'].value_counts().nlargest(10).index
df['Complaint Group'] = df['Complaint Type'].where(df['Complaint Type'].isin(top_categories), 'Other')
```

## Step 8 – Create Weekend Flag

```python
df['Is Weekend'] = (df['Created Date'].dt.weekday >= 5).astype(int)
```

## Step 9 – Preview Newly Created Features

```python
df[['Created Date', 'Closed Date', 'Month', 'Hour', 'Weekday', 'Resolution Hours', 'Is Open', 'Is Weekend', 'Complaint Group']].head()
```

## Step 10 – Analyze Complaints by Weekday

```python
df['Weekday'].value_counts()
```

## Step 11 – Visualize Complaints by Month

```python
df['Month'].value_counts().sort_index().plot(kind='bar')
plt.title('Complaints by Month')
plt.ylabel('Count')
plt.xlabel('Month')
plt.tight_layout()
plt.show()
```

## Key Takeaways

- Raw data often needs transformation before analysis
- Feature engineering improves interpretability
- Derived variables often reveal patterns hidden in raw fields
- Better features often matter more than more complex models
