# Day 13 – Outliers: Errors, Signals, or Reality?
## Practical Outlier Analysis using NYC 311 Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('nyc_311_sample.csv')
df.head()
```

```python
df['Created Date']=pd.to_datetime(df['Created Date'])
df['Closed Date']=pd.to_datetime(df['Closed Date'])
df['Resolution Hours']=(df['Closed Date']-df['Created Date']).dt.total_seconds()/3600
```

```python
df['Resolution Hours'].describe()
```

```python
plt.boxplot(df['Resolution Hours'].dropna())
plt.title('Resolution Time Outliers')
plt.show()
```

```python
Q1=df['Resolution Hours'].quantile(0.25)
Q3=df['Resolution Hours'].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
outliers=df[(df['Resolution Hours']<lower)|(df['Resolution Hours']>upper)]
outliers.head()
```

```python
print('Mean:',df['Resolution Hours'].mean())
print('Median:',df['Resolution Hours'].median())
```

```python
df_clean=df[df['Resolution Hours']>=0]
plt.hist(df_clean['Resolution Hours'].dropna(),bins=30)
plt.title('Distribution of Resolution Hours')
plt.show()
```

## Key Takeaways
- Outliers are not always errors
- Extreme values may reveal operational issues
- Compare mean and median before making conclusions
- Investigate anomalies before removing them
