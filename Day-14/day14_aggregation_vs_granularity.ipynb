# Day 14 – Aggregation vs Granularity
## How the Level of Detail Shapes Better Decisions

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("nyc_311_sample.csv")
df.head()
```

```python
df.info()
```

```python
df["Created Date"]=pd.to_datetime(df["Created Date"])
df["Year"]=df["Created Date"].dt.year
df["Month"]=df["Created Date"].dt.month
df["Day"]=df["Created Date"].dt.day
```

```python
len(df)
```

```python
borough=df.groupby("Borough").size().reset_index(name="Complaints")
borough
```

```python
borough.plot.bar(x="Borough",y="Complaints",legend=False,figsize=(8,4))
plt.title("Complaints by Borough")
plt.show()
```

```python
complaints=df.groupby("Complaint Type").size().sort_values(ascending=False)
complaints.head(10)
```

```python
complaints.head(10).plot.bar(figsize=(10,5))
plt.title("Top Complaint Types")
plt.show()
```

```python
monthly=df.groupby("Month").size()
monthly
```

```python
monthly.plot(marker="o")
plt.title("Monthly Complaint Trend")
plt.show()
```

```python
daily=df.groupby(df["Created Date"].dt.date).size()
daily.head()
```

```python
daily.plot(figsize=(10,4))
plt.title("Daily Complaint Trend")
plt.show()
```

```python
borough_month=df.groupby(["Borough","Month"]).size().reset_index(name="Complaints")
borough_month.head()
```

```python
pivot=pd.pivot_table(df,index="Borough",columns="Month",aggfunc="size",fill_value=0)
pivot
```

## Business Interpretation

- City summaries support strategic planning.
- Borough summaries support resource allocation.
- Daily summaries reveal operational events.
- Individual records support root-cause analysis.

## Key Takeaways

- Aggregation simplifies analysis.
- Granularity preserves detail.
- Different questions require different grouping levels.
- Avoid over-aggregation.

> **Aggregation tells us what happened. Granularity helps us understand why it happened.**
