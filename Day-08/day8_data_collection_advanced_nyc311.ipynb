## How Data Enters Analytical Systems

This notebook explores how operational datasets are created and how their collection mechanisms influence analysis.

Dataset used: **NYC 311 Service Requests**

Learning goals:
- Understand dataset structure
- Examine operational reporting fields
- Explore complaint distributions
- Analyze temporal reporting patterns
- Understand reporting bias in operational systems


## Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load Dataset
Download the NYC 311 dataset and update the file path if needed.

```python
df = pd.read_csv('311_Service_Requests_from_2010_to_Present.csv', low_memory=False)
df.head()
```

## Dataset Structure

```python
df.info()
```

## Key Fields Produced by Reporting Systems

```python
columns = ['Created Date','Complaint Type','Borough','Agency','Latitude','Longitude']
df[columns].head()
```

## Top Complaint Categories

```python
top_complaints = df['Complaint Type'].value_counts().head(10)
top_complaints
```

### Visualization – Top Complaint Categories

```python
top_complaints.plot(kind='bar')
plt.title('Top Complaint Categories')
plt.ylabel('Count')
plt.xlabel('Complaint Type')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Complaint Distribution by Borough

```python
borough_counts = df['Borough'].value_counts()
borough_counts
```

### Visualization – Borough Complaint Distribution

```python
borough_counts.plot(kind='bar')
plt.title('Complaints by Borough')
plt.ylabel('Number of Complaints')
plt.xlabel('Borough')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Time-Based Analysis
Operational datasets often contain time fields that reveal reporting behavior.

```python
df['Created Date'] = pd.to_datetime(df['Created Date'], errors='coerce')
df['Year'] = df['Created Date'].dt.year

yearly_counts = df['Year'].value_counts().sort_index()
yearly_counts
```

### Visualization – Complaints Over Time

```python
yearly_counts.plot()
plt.title('Complaint Reports Over Time')
plt.ylabel('Number of Complaints')
plt.xlabel('Year')
plt.tight_layout()
plt.show()
```

## Analytical Reflection

Because this dataset originates from a citizen reporting system, it reflects **reporting behavior**, not necessarily the full distribution of urban issues.

Factors influencing the data include:
- Awareness of reporting systems
- Accessibility of reporting platforms
- Categorization rules within the system
- Differences in citizen engagement

Understanding these factors is essential before drawing conclusions from the data.
