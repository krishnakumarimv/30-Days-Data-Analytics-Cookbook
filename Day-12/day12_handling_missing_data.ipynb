## Step 1 – Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Step 2 – Load Dataset

```python
df = pd.read_csv('nyc_311_sample.csv')
df.head()
```

## Step 3 – Identify Missing Values

```python
df.isnull().sum()
```

## Step 4 – Calculate Missing Percentages

```python
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_percent.sort_values(ascending=False)
```

## Step 5 – Visualize Missing Data

```python
missing_percent.sort_values(ascending=False).head(10).plot(kind='bar')
plt.title('Top Columns with Missing Data')
plt.ylabel('Missing Percentage')
plt.xlabel('Columns')
plt.tight_layout()
plt.show()
```

## Step 6 – Analyze Operational Meaning

```python
df['Closed Date'].isnull().sum()
```

## Step 7 – Create Missing Value Flags

```python
df['Missing Closed Date'] = df['Closed Date'].isnull().astype(int)
df[['Closed Date', 'Missing Closed Date']].head()
```

## Step 8 – Drop Rows Carefully

```python
df_clean = df.dropna(subset=['Complaint Type'])
df_clean.shape
```

## Step 9 – Fill Missing Values

```python
if 'Borough' in df.columns:
    mode_borough = df['Borough'].mode()[0]
    df['Borough'] = df['Borough'].fillna(mode_borough)

if 'Latitude' in df.columns:
    median_lat = df['Latitude'].median()
    df['Latitude'] = df['Latitude'].fillna(median_lat)

df.isnull().sum()
```

## Step 10 – Final Dataset Check

```python
df.info()
```

## Key Takeaways

- Missing data is often analytically meaningful
- Blind deletion may distort conclusions
- Different variables require different handling strategies
- Missingness can reveal operational issues and process gaps
