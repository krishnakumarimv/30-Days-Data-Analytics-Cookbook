# Day 17 – Correlation vs. Causation

## Why Relationships in Data Can Be Misleading

```
import pandas as pd
import matplotlib.pyplot as plt
```

```
df=pd.read_csv('data/nyc_311_sample.csv')
df.head()
```

```
df['Created Date']=pd.to_datetime(df['Created Date'])
df['Closed Date']=pd.to_datetime(df['Closed Date'])
df['Resolution Hours']=(df['Closed Date']-df['Created Date']).dt.total_seconds()/3600
```

```
numeric=df.select_dtypes(include='number')
corr=numeric.corr()
corr
```

```
plt.imshow(corr,cmap='coolwarm')
plt.colorbar()
plt.show()
```

```
plt.scatter(df['Resolution Hours'],df.index,alpha=0.4)
plt.show()
```

### Key Takeaways
- Correlation does not imply causation.
- Consider confounding variables.
- Validate with domain knowledge.
