Dataset Example: NYC 311 Service Requests

## Step 1 – Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Step 2 – Load Dataset

```python
df = pd.read_csv('nyc_311_sample.csv')
df.head()
```

## Step 3 – Inspect Dataset

```python
df.shape
```

```python
df.info()
```

## Step 4 – Missing Values

```python
missing = df.isnull().sum()
missing.sort_values(ascending=False)
```

## Step 5 – Duplicate Detection

```python
df.duplicated().sum()
```

## Step 6 – Remove Duplicates

```python
df = df.drop_duplicates()
df.shape
```

## Step 7 – Date Conversion

```python
df['Created Date'] = pd.to_datetime(df['Created Date'], errors='coerce')
df['Closed Date'] = pd.to_datetime(df['Closed Date'], errors='coerce')
```

## Step 8 – Handle Missing Borough Values

```python
df['Borough'] = df['Borough'].fillna('Unknown')
```

## Step 9 – Complaint Category Check

```python
df['Complaint Type'].value_counts().head(10)
```

## Step 10 – Data Quality Visualization

```python
missing.head(10).plot(kind='bar')
plt.title('Missing Values by Column')
plt.ylabel('Count')
plt.show()
```

## Key Takeaways
- Real-world data requires cleaning
- Missing values distort analysis
- Duplicate records affect counts
- Data validation is essential before analytics
