# Day 15 – Descriptive Statistics
## Understanding Data Before Building Models

```
import pandas as pd
import matplotlib.pyplot as plt
```

```
df=pd.read_csv("data/nyc_311_sample.csv")
df.head()
```

```
print(df.shape)
df.info()
```

```
print(df.columns)
```

```
df["Created Date"]=pd.to_datetime(df["Created Date"])
```

```
df.describe()
```

```
df["Resolution Hours"].mean()
```

```
df["Resolution Hours"].median()
```

```
df["Resolution Hours"].mode()
```

```
df["Resolution Hours"].std()
```

```
df["Resolution Hours"].quantile([0.25,0.5,0.75])
```

```
complaints=df["Complaint Type"].value_counts()
complaints.head(10)
```

```
borough=df["Borough"].value_counts()
borough
```

```
df["Resolution Hours"].plot.hist(bins=20)
plt.show()
```

```
df.boxplot(column="Resolution Hours")
plt.show()
```

```
complaints.head(10).plot.bar()
plt.show()
```

```
borough.plot.bar()
plt.show()
```

## Business Interpretation
Descriptive statistics summarize and explain the characteristics of a dataset before deeper analysis.

## Key Takeaways
- Understand your data first.
- Compare mean, median and mode.
- Measure variability.
- Analyze frequencies before modeling.

> **Before asking what will happen next, understand what is happening now.**
