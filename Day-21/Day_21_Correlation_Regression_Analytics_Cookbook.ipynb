## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import pearsonr
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

np.random.seed(42)
```

## 2. Create a Synthetic Public Grievance Dataset

We examine whether monthly complaint volume is related to average resolution time.

```python
months = pd.date_range("2024-01-01", periods=24, freq="MS")

complaints = np.array([
    420, 450, 470, 510, 530, 560, 590, 610,
    640, 680, 700, 730, 760, 790, 820, 850,
    890, 910, 940, 970, 1000, 1040, 1080, 1120
])

resolution_days = (
    2.8 + 0.006 * complaints +
    np.random.normal(0, 0.35, len(complaints))
)

df = pd.DataFrame({
    "month": months,
    "complaints": complaints,
    "resolution_days": resolution_days
})

df.head()
```

## 3. Explore the Relationship

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["complaints"], df["resolution_days"])
plt.xlabel("Monthly Complaint Volume")
plt.ylabel("Average Resolution Time (Days)")
plt.title("Complaint Volume vs Resolution Time")
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Pearson Correlation

Pearson's `r` ranges from -1 to +1.

- `+1` = perfect positive linear relationship
- `0` = no linear relationship
- `-1` = perfect negative linear relationship

**Correlation does not prove causation.**

```python
r, p_value = pearsonr(
    df["complaints"],
    df["resolution_days"]
)

print(f"Pearson correlation (r): {r:.3f}")
print(f"P-value: {p_value:.6f}")
```

## 5. Simple Linear Regression

**Y = β₀ + β₁X + ε**

Where `Y` is the outcome, `X` is the predictor, `β₀` is the intercept, `β₁` is the slope, and `ε` is the error.

```python
X = df[["complaints"]]
y = df["resolution_days"]

model = LinearRegression()
model.fit(X, y)

intercept = model.intercept_
slope = model.coef_[0]

print(f"Intercept: {intercept:.3f}")
print(f"Slope: {slope:.5f}")
```

## 6. Regression Equation and Predictions

```python
df["predicted_resolution_days"] = model.predict(X)

print(
    f"Resolution Days = {intercept:.3f} + "
    f"{slope:.5f} × Complaints"
)

df.head()
```

## 7. Visualize the Regression Line

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["complaints"], df["resolution_days"], label="Observed")
plt.plot(df["complaints"], df["predicted_resolution_days"], label="Regression line")
plt.xlabel("Monthly Complaint Volume")
plt.ylabel("Average Resolution Time (Days)")
plt.title("Linear Regression")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 8. Evaluate the Model

- **R²** – proportion of variation explained
- **MAE** – average absolute prediction error
- **RMSE** – square root of mean squared prediction error

```python
y_pred = model.predict(X)

r2 = r2_score(y, y_pred)
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print(f"R²: {r2:.3f}")
print(f"MAE: {mae:.3f} days")
print(f"RMSE: {rmse:.3f} days")
```

## 9. Residual Analysis

**Residual = Actual − Predicted**

Residuals help identify systematic model errors.

```python
df["residual"] = df["resolution_days"] - df["predicted_resolution_days"]

plt.figure(figsize=(8, 5))
plt.scatter(df["predicted_resolution_days"], df["residual"])
plt.axhline(0, linestyle="--")
plt.xlabel("Predicted Resolution Time (Days)")
plt.ylabel("Residual")
plt.title("Residual Plot")
plt.grid(True, alpha=0.3)
plt.show()
```

## 10. Make a New Prediction

```python
future_complaints = pd.DataFrame({"complaints": [1200]})
prediction = model.predict(future_complaints)[0]

print(f"Predicted average resolution time: {prediction:.2f} days")
```

## 11. Correlation Matrix

```python
analysis_df = df[["complaints", "resolution_days", "predicted_resolution_days"]]
print(analysis_df.corr(numeric_only=True).round(3))
```

## 12. Simple vs Multiple Regression

Real governance problems often depend on several variables: complaint volume, category, zone, workforce availability, severity, seasonality and department workload.

Multiple regression extends the model to several predictors:

**Y = β₀ + β₁X₁ + β₂X₂ + ... + βₖXₖ + ε**

## 13. Regression Assumptions

Investigate:
1. Linearity
2. Independence
3. Homoscedasticity
4. Residual distribution
5. Multicollinearity
6. Outliers and influential observations

A high R² alone does not guarantee a good model.

## 14. Public Governance Applications

Correlation and regression can support:
- Complaint workload analysis
- Resolution-time prediction
- Demand forecasting
- Service-level monitoring
- Resource allocation
- Ward/zone performance analysis
- Identifying operational drivers
- Early-warning systems

**Governance principle:** statistical relationships should support investigation and decision-making, not be treated as automatic proof of causality.

## 15. Key Takeaway

**Correlation:** What variables move together?

**Regression:** Can we quantify the relationship and use it for prediction?
